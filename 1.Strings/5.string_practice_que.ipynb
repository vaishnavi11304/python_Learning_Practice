#### Q1. Write a Python program to reverse a string.

```python
a="vaishnavi aher , 1,2,3,4,5,6,7"
b=a[::-1]
print(b)
```

#### Q2. Write a Python program to check whether a string is a palindrome.

```python
word = "nayan"
if word == word[::-1]:
    print("palindrome")
else:
    print("not palindrome")
```

#### Q3. Write a Python program to find the length of a string without using len().

```python
sentence="hi i'm vaishnavi  from pune"
count=0
for char in sentence:
    count+=1
print(count)
```

#### Q4. Write a Python program to count the number of vowels in a string.

```python
a="a,b,c,d,e,o,u,i,a,e,i,o,u,a,aa,a,"
for char in "aeiou":
    print(char ,"=", a.count(char))
```
